# ⚖️ 4. 통합 벤치마크 평가 (evaluate_all.py)
Standard Metrics와 K-PatientBench를 동시에 평가하고 시각화합니다.

````python
import os, json, torch, numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from tqdm import tqdm
from openai import OpenAI
from unsloth import FastLanguageModel
from dotenv import load_dotenv
from math import pi

load_dotenv()
OPENAI_CLIENT = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
READ_TOKEN = os.getenv("HUGGINGFACE_API_KEY_FOR_READ_ONLY")
MODELS = {"Baseline (SFT)": "sft_model_a", "Ours (DPO)": "final_dpo_model_b"}

# --- 1. 벤치마크 데이터 생성 ---
EVAL_DATA = [
    # Track A: Standard
    {
        "group": "Standard",
        "metric": "Prompt-to-Line",
        "phase": "All",
        "persona": "김철수, 32세, 작가",
        "query": "본인 소개 부탁해요.",
        "criteria": "32세 작가 설정 유지",
    },
    {
        "group": "Standard",
        "metric": "Line-to-Line",
        "phase": "All",
        "persona": "방금 밥 먹었다고 함",
        "query": "식사는 하셨나요?",
        "criteria": "먹었다고 답해야 함",
    },
    # Track B: K-PatientBench
    {
        "group": "K-Patient",
        "metric": "Resistance (S1)",
        "phase": "Early",
        "persona": "상담 불신, 방어적",
        "query": "오시는 길 힘들지 않으셨나요?",
        "criteria": "단답형, 귀찮음, 냉소적 태도",
    },
    {
        "group": "K-Patient",
        "metric": "Resistance (S1)",
        "phase": "Early",
        "persona": "무기력, 외부 탓",
        "query": "요즘 가장 힘든 점은?",
        "criteria": "환경 탓, 남 탓, 회피",
    },
    {
        "group": "K-Patient",
        "metric": "Change_Talk (S5)",
        "phase": "Late",
        "persona": "변화 의지 있음",
        "query": "지난주 산책은 어떠셨나요?",
        "criteria": "시도함(Ability), 의지(Desire) 표현",
    },
    {
        "group": "K-Patient",
        "metric": "Change_Talk (S5)",
        "phase": "Late",
        "persona": "통찰 획득",
        "query": "앞으로의 목표는?",
        "criteria": "구체적이고 긍정적인 계획(Plan)",
    },
]


# --- 2. 추론 및 채점 ---
def run_eval():
    results = []
    alpaca = "Below is an instruction...\n\n### Instruction:\n당신은 내담자입니다. 상황에 맞춰 대답하세요.\n\n### Input:\n[상황: {phase}]\n설정: {persona}\n상담사: {query}\n\n### Response:\n"

    # 모델별 추론
    for name, path in MODELS.items():
        print(f"\n🚀 Testing {name}...")
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name=path, max_seq_length=2048, load_in_4bit=True, token=READ_TOKEN
        )
        FastLanguageModel.for_inference(model)

        for item in tqdm(EVAL_DATA):
            prompt = alpaca.format(
                phase=item["phase"], persona=item["persona"], query=item["query"]
            )
            inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
            outputs = model.generate(**inputs, max_new_tokens=150, use_cache=True)
            res = (
                tokenizer.batch_decode(outputs, skip_special_tokens=True)[0]
                .split("### Response:\n")[-1]
                .strip()
            )

            # 채점 (Judge)
            judge_prompt = f"평가 기준: {item['metric']} / 정답 기준: {item['criteria']} / 답변: {res}\n\n1~5점 평가 및 이유를 JSON으로 출력: {{'score': 점수, 'reason': '이유'}}"
            try:
                api_res = OPENAI_CLIENT.chat.completions.create(
                    model="gpt-4o-mini",
                    messages=[{"role": "user", "content": judge_prompt}],
                    response_format={"type": "json_object"},
                )
                score = json.loads(api_res.choices[0].message.content)
                results.append(
                    {**item, "model": name, "response": res, "score": score["score"]}
                )
            except:
                results.append({**item, "model": name, "response": res, "score": 0})

        del model, tokenizer
        torch.cuda.empty_cache()

    return pd.DataFrame(results)


# --- 3. 시각화 (Radar Chart) ---
def visualize(df):
    df_k = df[df["group"] == "K-Patient"]
    pivot = df_k.groupby(["model", "metric"])["score"].mean().unstack()
    categories = list(pivot.columns)

    angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
    angles += angles[:1]

    fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
    plt.xticks(angles[:-1], categories)

    colors = {"Baseline (SFT)": "grey", "Ours (DPO)": "red"}
    for model in pivot.index:
        values = pivot.loc[model].tolist()
        values += values[:1]
        ax.plot(angles, values, linewidth=2, label=model, color=colors[model])
        ax.fill(angles, values, color=colors[model], alpha=0.25)

    plt.legend(loc="upper right", bbox_to_anchor=(0.1, 0.1))
    plt.savefig("final_radar_chart.png", dpi=300)
    print("📊 Radar Chart 저장 완료")


if __name__ == "__main__":
    df = run_eval()
    df.to_csv("final_results.csv", index=False)
    visualize(df)
    print("🎉 전체 프로젝트 완료!")
````

````python
import os
import json
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from openai import OpenAI
from unsloth import FastLanguageModel
from dotenv import load_dotenv
from math import pi

# 1. 환경 설정
load_dotenv()
OPENAI_CLIENT = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
READ_TOKEN = os.getenv("HUGGINGFACE_API_KEY_FOR_READ_ONLY")

# 🚨 [수정됨] 로컬 경로 대신 Hugging Face ID 사용
MODELS = {
    "Baseline (SFT)": "hyunNus/Woooly-SFT-70B",
    "Ours (DPO)": "hyunNus/Woooly-SFT+DPO-70B",
}

# 2. 평가 데이터 (Standard + K-Patient)
EVAL_DATA = [
    # --- Track A: Standard Metrics (기존 연구) ---
    {
        "group": "Standard",
        "metric": "Prompt-to-Line",
        "phase": "All",
        "persona": "김철수, 32세, 작가",
        "query": "본인 소개 부탁해요.",
        "criteria": "32세 작가 설정 유지",
    },
    {
        "group": "Standard",
        "metric": "Line-to-Line",
        "phase": "All",
        "persona": "방금 밥 먹었다고 함",
        "query": "식사는 하셨나요?",
        "criteria": "먹었다고 답해야 함 (문맥 일치)",
    },
    # --- Track B: K-PatientBench (제안 연구) ---
    {
        "group": "K-Patient",
        "metric": "Resistance",
        "phase": "Early",
        "persona": "우울증 초기, 상담 불신",
        "query": "오시는 길 힘들지 않으셨나요?",
        "criteria": "단답형, 귀찮음, 냉소적 태도",
    },
    {
        "group": "K-Patient",
        "metric": "Resistance",
        "phase": "Early",
        "persona": "무기력, 외부 탓",
        "query": "요즘 가장 힘든 점은?",
        "criteria": "환경 탓, 남 탓, 회피",
    },
    {
        "group": "K-Patient",
        "metric": "Change_Talk",
        "phase": "Late",
        "persona": "변화 의지 있음",
        "query": "지난주 산책은 어떠셨나요?",
        "criteria": "시도함(Ability), 의지(Desire) 표현",
    },
    {
        "group": "K-Patient",
        "metric": "Change_Talk",
        "phase": "Late",
        "persona": "통찰 획득",
        "query": "앞으로의 목표는?",
        "criteria": "구체적이고 긍정적인 계획(Plan)",
    },
]


# 3. 추론 함수
def run_inference(name, path):
    print(f"\n🚀 Testing {name} ({path})...")
    try:
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name=path, max_seq_length=2048, load_in_4bit=True, token=READ_TOKEN
        )
    except Exception as e:
        print(f"❌ 모델 로드 실패: {path}. 경로를 확인하세요. ({e})")
        return []

    FastLanguageModel.for_inference(model)

    results = []
    alpaca = "Below is an instruction...\n\n### Instruction:\n당신은 내담자입니다. 상황에 맞춰 대답하세요.\n\n### Input:\n[상황: {phase}]\n설정: {persona}\n상담사: {query}\n\n### Response:\n"

    for item in tqdm(EVAL_DATA, desc=f"Generating ({name})"):
        prompt = alpaca.format(
            phase=item["phase"], persona=item["persona"], query=item["query"]
        )
        inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
        outputs = model.generate(**inputs, max_new_tokens=150, use_cache=True)
        res = (
            tokenizer.batch_decode(outputs, skip_special_tokens=True)[0]
            .split("### Response:\n")[-1]
            .strip()
        )
        results.append({**item, "model": name, "response": res})

    del model, tokenizer
    torch.cuda.empty_cache()
    return results


# 4. 채점 함수 (JSON 파싱 보완)
def run_judge(results):
    print("\n⚖️ Judging...")
    scored_results = []
    for r in tqdm(results):
        prompt = f"평가 기준: {r['metric']} / 정답 기준: {r['criteria']} / 답변: {r['response']}\n\n1~5점 평가 및 이유를 JSON으로 출력: {{'score': 점수, 'reason': '이유'}}"
        try:
            api_res = OPENAI_CLIENT.chat.completions.create(
                model="gpt-4o-mini",
                messages=[{"role": "user", "content": prompt}],
                response_format={"type": "json_object"},
            )
            content = api_res.choices[0].message.content

            # JSON 파싱 강화 (Markdown 제거)
            if content.startswith("```json"):
                content = content.replace("```json", "").replace("```", "")

            score_data = json.loads(content)
            r["score"] = score_data["score"]
            r["reason"] = score_data["reason"]
        except Exception as e:
            print(f"⚠️ 채점 에러: {e}")
            r["score"] = 0
            r["reason"] = "Error"
        scored_results.append(r)
    return pd.DataFrame(scored_results)


# 5. 시각화 (Radar + Bar)
def visualize(df):
    # [1] Radar Chart (K-PatientBench)
    df_k = df[df["group"] == "K-Patient"]
    if not df_k.empty:
        pivot = df_k.groupby(["model", "metric"])["score"].mean().unstack()
        categories = list(pivot.columns)
        angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
        angles += angles[:1]

        fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
        plt.xticks(angles[:-1], categories, size=12)
        colors = {"Baseline (SFT)": "grey", "Ours (DPO)": "red"}

        for model in pivot.index:
            val = pivot.loc[model].tolist()
            val += val[:1]
            ax.plot(
                angles, val, linewidth=2, label=model, color=colors.get(model, "blue")
            )
            ax.fill(angles, val, color=colors.get(model, "blue"), alpha=0.25)

        plt.title("Dynamic Persona Capability (K-PatientBench)", y=1.1, weight="bold")
        plt.legend(loc="upper right", bbox_to_anchor=(0.1, 0.1))
        plt.savefig("final_radar_chart.png", dpi=300)
        print("📊 Radar Chart 저장 완료")

    # [2] Bar Chart (Standard Metrics)
    df_s = df[df["group"] == "Standard"]
    if not df_s.empty:
        plt.figure(figsize=(8, 5))
        sns.barplot(
            data=df_s, x="metric", y="score", hue="model", palette=["grey", "red"]
        )
        plt.title("Standard Consistency Metrics", weight="bold")
        plt.ylim(0, 5.5)
        plt.savefig("final_bar_chart.png", dpi=300)
        print("📊 Bar Chart 저장 완료")


# 6. 메인 실행
if __name__ == "__main__":
    all_res = []
    for name, path in MODELS.items():
        all_res.extend(run_inference(name, path))

    if all_res:
        final_df = run_judge(all_res)
        final_df.to_csv("final_results.csv", index=False, encoding="utf-8-sig")
        visualize(final_df)
        print("\n🎉 전체 평가 완료! CSV와 이미지 파일을 확인하세요.")
        print(final_df.groupby(["model", "group", "metric"])["score"].mean())
    else:
        print("❌ 추론된 결과가 없습니다. 모델 경로를 확인하세요.")
````

````python
import os
import json
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from openai import OpenAI
from unsloth import FastLanguageModel
from dotenv import load_dotenv
from math import pi

# 1. 환경 설정
load_dotenv()
OPENAI_CLIENT = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
READ_TOKEN = os.getenv("HUGGINGFACE_API_KEY_FOR_READ_ONLY")

# 모델 설정
MODELS = {
    "Baseline (SFT)": "hyunNus/Woooly-SFT-70B", 
    "Ours (DPO)": "hyunNus/Woooly-SFT-DPO-70B"
}

# 2. 평가 데이터 (동일)
EVAL_DATA = [
    # Track A: Standard
    {"group": "Standard", "metric": "Prompt Consistency", "phase": "All", "persona": "김철수, 32세, 작가", "query": "본인 소개 부탁해요.", "criteria": "32세 작가 설정 유지"},
    {"group": "Standard", "metric": "Line Consistency", "phase": "All", "persona": "방금 밥 먹었다고 함", "query": "식사는 하셨나요?", "criteria": "먹었다고 답해야 함"},
    # Track B: K-PatientBench (Session 1 - 저항)
    {"group": "K-Patient", "metric": "Initial Resistance (S1)", "phase": "Early", "persona": "우울증 초기, 상담 불신", "query": "오시는 길 힘들지 않으셨나요?", "criteria": "방어적 태도 유지"},
    {"group": "K-Patient", "metric": "Initial Resistance (S1)", "phase": "Early", "persona": "무기력, 외부 탓", "query": "요즘 가장 힘든 점은?", "criteria": "환경/남 탓, 회피"},
    # Track B: K-PatientBench (Session 5 - 변화)
    {"group": "K-Patient", "metric": "Change Talk (S5)", "phase": "Late", "persona": "변화 의지 있음", "query": "지난주 산책은 어떠셨나요?", "criteria": "시도함(Ability), 의지(Desire)"},
    {"group": "K-Patient", "metric": "Change Talk (S5)", "phase": "Late", "persona": "통찰 획득", "query": "앞으로의 목표는?", "criteria": "구체적이고 긍정적인 계획(Plan)"}
]

# 3. 추론 함수 (동일)
def run_inference(name, path):
    print(f"\n🚀 Testing {name} ({path})...")
    try:
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name=path, max_seq_length=2048, load_in_4bit=True, token=READ_TOKEN
        )
    except Exception as e:
        print(f"❌ 모델 로드 실패: {e}")
        return []

    FastLanguageModel.for_inference(model)
    results = []
    alpaca = "Below is an instruction...\n\n### Instruction:\n당신은 내담자입니다. 상황에 맞춰 대답하세요.\n\n### Input:\n[상황: {phase}]\n설정: {persona}\n상담사: {query}\n\n### Response:\n"
    
    for item in tqdm(EVAL_DATA, desc=f"Generating ({name})"):
        prompt = alpaca.format(phase=item['phase'], persona=item['persona'], query=item['query'])
        inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
        outputs = model.generate(**inputs, max_new_tokens=150, use_cache=True)
        res = tokenizer.batch_decode(outputs, skip_special_tokens=True)[0].split("### Response:\n")[-1].strip()
        results.append({**item, "model": name, "response": res})
    
    del model, tokenizer
    torch.cuda.empty_cache()
    return results

# ---------------------------------------------------------
# 4. [핵심 수정] 고도화된 심사 프롬프트 (Prompt Engineering)
# ---------------------------------------------------------
JUDGE_SYSTEM_PROMPT = """
당신은 심리 상담 AI 모델을 평가하는 '임상 심리 수퍼바이저(Clinical Supervisor)'입니다.
아래 제공된 [평가 가이드라인]에 따라, AI가 연기하는 내담자(Client)의 답변을 냉철하게 평가하십시오.
"""

JUDGE_USER_PROMPT = """
### [평가 가이드라인]
1. **Prompt Consistency (페르소나 일치):** 답변이 주어진 나이, 직업, 성격 설정과 모순되지 않는지 확인하십시오.
2. **Resistance (저항 - 초기 단계):** 상담 초기라면, 내담자는 방어적이고, 단답형이며, 상담사를 불신하거나 문제의 원인을 외부로 돌려야(Projection) 합니다. 너무 협조적이면 감점입니다.
3. **Change Talk (변화 대화 - 후기 단계):** 상담 후기라면, 내담자는 변화에 대한 욕구(Desire), 능력(Ability), 이유(Reason), 필요(Need)를 표현해야 합니다. (KMI 데이터셋 기준)

### [입력 정보]
- 평가 항목: {metric} (그룹: {group})
- 상황 설정: {persona} (단계: {phase})
- 상담사 질문: "{query}"
- **정답 기준 (Criteria):** {criteria}

### [AI 답변]
"{response}"

### [채점 기준 (1~5점)]
- **5점 (완벽):** 정답 기준을 완벽하게 충족하며, 해당 단계(Phase)에 맞는 심리적 뉘앙스(저항 또는 변화)를 자연스럽게 표현함.
- **3점 (보통):** 내용은 맞지만, 태도가 너무 기계적이거나 심리적 깊이가 부족함.
- **1점 (실패):** 설정과 모순되거나(Hallucination), 초기인데 너무 친절하거나 후기인데 변화 의지가 없음.

**반드시 아래 JSON 형식으로만 출력하시오.**
{{ "score": 점수, "reason": "평가 이유 한 줄 요약" }}
"""

def run_judge(results):
    print("\n⚖️ Judging (with Advanced Prompt)...")
    scored_results = []
    for r in tqdm(results):
        try:
            full_prompt = JUDGE_USER_PROMPT.format(
                metric=r['metric'], group=r['group'], 
                persona=r['persona'], phase=r['phase'],
                query=r['query'], criteria=r['criteria'], 
                response=r['response']
            )
            
            api_res = OPENAI_CLIENT.chat.completions.create(
                model="gpt-4o-mini",
                messages=[
                    {"role": "system", "content": JUDGE_SYSTEM_PROMPT},
                    {"role": "user", "content": full_prompt}
                ],
                response_format={"type": "json_object"},
                temperature=0 # 평가는 일관성이 중요하므로 0
            )
            
            content = api_res.choices[0].message.content
            if content.startswith("```json"): content = content.replace("```json", "").replace("```", "")
            score_data = json.loads(content)
            
            r['score'] = score_data['score']
            r['reason'] = score_data['reason']
        except Exception as e:
            print(f"⚠️ Error: {e}")
            r['score'] = 0
            r['reason'] = "Error"
        scored_results.append(r)
    return pd.DataFrame(scored_results)

# 5. 시각화 (동일)
def visualize(df):
    # [1] Radar Chart
    df_k = df[df['group'] == 'K-Patient']
    if not df_k.empty:
        pivot = df_k.groupby(['model', 'metric'])['score'].mean().unstack()
        # 정렬: 저항 -> 변화 순서로 보기 좋게
        cols = ['Initial Resistance (S1)', 'Change Talk (S5)'] 
        categories = [c for c in cols if c in pivot.columns]
        
        angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
        angles += angles[:1]
        
        fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
        plt.xticks(angles[:-1], categories, size=12, weight='bold')
        colors = {'Baseline (SFT)': '#999999', 'Ours (DPO)': '#E63946'} # 회색 vs 빨강
        
        for model in pivot.index:
            val = pivot.loc[model, categories].tolist()
            val += val[:1]
            ax.plot(angles, val, linewidth=2, label=model, color=colors.get(model, 'blue'))
            ax.fill(angles, val, color=colors.get(model, 'blue'), alpha=0.2)
        
        plt.title("Dynamic Persona Alignment (K-PatientBench)", y=1.1, weight='bold')
        plt.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
        plt.savefig("final_radar_chart.png", dpi=300, bbox_inches='tight')
        print("📊 Radar Chart 저장 완료")

    # [2] Bar Chart
    df_s = df[df['group'] == 'Standard']
    if not df_s.empty:
        plt.figure(figsize=(8, 5))
        sns.barplot(data=df_s, x="metric", y="score", hue="model", palette=["#999999", "#E63946"])
        plt.title("Standard Consistency Metrics", weight='bold')
        plt.ylim(0, 5.5)
        plt.ylabel("Score (1-5)")
        plt.savefig("final_bar_chart.png", dpi=300)
        print("📊 Bar Chart 저장 완료")

# 6. 메인 실행
if __name__ == "__main__":
    all_res = []
    for name, path in MODELS.items():
        all_res.extend(run_inference(name, path))
    
    if all_res:
        final_df = run_judge(all_res)
        final_df.to_csv("final_results.csv", index=False, encoding="utf-8-sig")
        visualize(final_df)
        print("\n🎉 평가 완료! CSV와 그래프를 확인하세요.")
        print(final_df.groupby(["model", "group", "metric"])["score"].mean())
````

# 🧬 벤치마크 데이터 생성

````python
import json
import os
import random
from tqdm import tqdm
from openai import OpenAI
from dotenv import load_dotenv

# 1. 환경 설정
load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

INPUT_FILE = "kmi.json"
OUTPUT_FILE = "integrated_benchmark.json"
TARGET_COUNT = 50  # 생성할 페르소나 개수 (총 문항 수는 이보다 많음)

# 2. 벤치마크 생성 프롬프트 (GPT-4o-mini)
GEN_PROMPT = """
당신은 임상 심리 벤치마크 출제자입니다.
제공된 [상담 데이터]를 분석하여, AI 상담 내담자 모델을 평가하기 위한 '테스트 케이스 세트'를 생성하세요.

[입력 데이터]
- 카테고리: {category}
- 대화 요약: {dialogue_text}

[생성 요청]
다음 JSON 형식으로 4개의 테스트 케이스를 생성하세요:

1. **persona**: 내담자의 나이, 직업, 핵심 증상, 성격을 1문장으로 요약.
2. **cases**: [
    {{
        "group": "Standard", "metric": "Prompt-to-Line", "phase": "All",
        "query": "(페르소나 확인 질문, 예: 자기소개)", "criteria": "(이름/직업 등 설정 유지)"
    }},
    {{
        "group": "K-Patient", "metric": "Resistance", "phase": "Early",
        "query": "(상담 초기 탐색 질문)", "criteria": "방어적, 단답형, 외부 탓, 불신 (저항)"
    }},
    {{
        "group": "K-Patient", "metric": "Change Talk", "phase": "Late",
        "query": "(변화나 통찰 유도 질문)", "criteria": "변화 의지(Desire/Ability), 통찰, 구체적 계획 (수용)"
    }}
]

**반드시 JSON 포맷만 출력하세요.**
"""

def generate_bench_case(category, dialogue):
    # 대화 내용을 텍스트로 변환 (앞부분만 참조)
    dialogue_text = "\n".join([f"{t['role']}: {t.get('utterance_ko','')}" for t in dialogue[:8]])
    
    try:
        res = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": GEN_PROMPT.format(category=category, dialogue_text=dialogue_text)}],
            response_format={"type": "json_object"}
        )
        return json.loads(res.choices[0].message.content)
    except:
        return None

def main():
    if not os.path.exists(INPUT_FILE):
        print(f"❌ {INPUT_FILE} 파일이 없습니다.")
        return

    with open(INPUT_FILE, 'r', encoding='utf-8') as f:
        raw_data = json.load(f)
    
    # 다양한 카테고리가 섞이도록 셔플
    random.shuffle(raw_data)
    
    final_benchmark = []
    
    print(f"🚀 벤치마크 데이터 확장 시작 (목표 페르소나: {TARGET_COUNT}개)...")
    
    count = 0
    for sample in tqdm(raw_data):
        if count >= TARGET_COUNT: break
            
        dlg = sample.get('dialogue', [])
        cat = sample.get('category_ko', '일반')
        
        if len(dlg) < 4: continue
        
        # GPT-4o-mini로 케이스 생성
        data = generate_bench_case(cat, dlg)
        
        if data and 'persona' in data and 'cases' in data:
            persona = data['persona']
            for case in data['cases']:
                # 공통 필드 추가
                case['persona'] = persona
                final_benchmark.append(case)
            count += 1

    # 저장
    with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
        json.dump(final_benchmark, f, indent=2, ensure_ascii=False)
        
    print(f"✅ 벤치마크 생성 완료! 총 {len(final_benchmark)}개 문항이 '{OUTPUT_FILE}'에 저장되었습니다.")

if __name__ == "__main__":
    main()
````

# ⚖️ 최종 통합 평가 코드 (evaluate_integrated.py)

````python
import os
import json
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from openai import OpenAI
from unsloth import FastLanguageModel
from dotenv import load_dotenv
from math import pi

# 1. 환경 설정
load_dotenv()
OPENAI_CLIENT = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
READ_TOKEN = os.getenv("HUGGINGFACE_API_KEY_FOR_READ_ONLY")

# 🚨 평가 대상 모델 (Hugging Face ID)
MODELS = {
    "Baseline (SFT)": "hyunNus/Woooly-SFT-70B", 
    "Ours (DPO)": "hyunNus/Woooly-SFT-DPO-70B"
}

# 2. 추론 함수 (메모리 관리 포함)
def run_inference(name, path, dataset):
    print(f"\n🚀 Testing {name} ({path})...")
    try:
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name=path, max_seq_length=2048, load_in_4bit=True, token=READ_TOKEN
        )
    except Exception as e:
        print(f"❌ 모델 로드 실패: {path}. ({e})")
        return []

    FastLanguageModel.for_inference(model)
    
    results = []
    alpaca = "Below is an instruction...\n\n### Instruction:\n당신은 내담자입니다. 상황에 맞춰 대답하세요.\n\n### Input:\n[상황: {phase}]\n설정: {persona}\n상담사: {query}\n\n### Response:\n"
    
    for item in tqdm(dataset, desc=f"Generating ({name})"):
        # 모델에 입력할 프롬프트 구성
        phase_str = item['phase']
        if "Early" in phase_str or "Session 1" in phase_str: phase_input = "Session 1 (초기 저항)"
        elif "Late" in phase_str or "Session 5" in phase_str: phase_input = "Session 5 (변화/통찰)"
        else: phase_input = "일반 상담"

        prompt = alpaca.format(phase=phase_input, persona=item['persona'], query=item['query'])
        
        inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
        outputs = model.generate(**inputs, max_new_tokens=150, use_cache=True)
        res = tokenizer.batch_decode(outputs, skip_special_tokens=True)[0].split("### Response:\n")[-1].strip()
        
        results.append({**item, "model": name, "response": res})
    
    del model, tokenizer
    torch.cuda.empty_cache()
    return results

# 3. 채점 함수 (LLM-as-a-Judge)
JUDGE_PROMPT = """
당신은 임상 심리 시뮬레이션 평가자입니다. 아래 기준에 따라 AI 답변을 1~5점으로 평가하세요.

[평가 정보]
- 평가 항목: {metric} (그룹: {group})
- 상황 설정: {persona} ({phase})
- 상담사 질문: "{query}"
- **정답 기준:** {criteria}

[AI 답변]
"{response}"

[채점 가이드]
- 5점: 정답 기준을 완벽 충족하며, 해당 단계(저항/변화)의 심리적 뉘앙스를 잘 표현함.
- 3점: 기준을 어느 정도 충족하나, 표현이 단순하거나 기계적임.
- 1점: 기준과 반대되는 태도(예: 저항해야 하는데 친절함)를 보이거나 환각 발생.

JSON으로 출력: {{ "score": 점수, "reason": "이유" }}
"""

def run_judge(results):
    print("\n⚖️ Judging...")
    scored = []
    for r in tqdm(results):
        try:
            prompt = JUDGE_PROMPT.format(**r)
            res = OPENAI_CLIENT.chat.completions.create(
                model="gpt-4o-mini", messages=[{"role":"user","content":prompt}], response_format={"type":"json_object"}
            )
            score_data = json.loads(res.choices[0].message.content)
            r['score'] = score_data['score']
            r['reason'] = score_data['reason']
        except:
            r['score'] = 0
            r['reason'] = "Error"
        scored.append(r)
    return pd.DataFrame(scored)

# 4. 시각화 함수
def visualize(df):
    # A. Radar Chart (K-PatientBench)
    df_k = df[df['group'] == 'K-Patient']
    if not df_k.empty:
        pivot = df_k.groupby(['model', 'metric'])['score'].mean().unstack()
        cols = ['Resistance', 'Change Talk'] # 순서 지정
        categories = [c for c in cols if c in pivot.columns]
        
        if categories:
            angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
            angles += angles[:1]
            
            fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
            plt.xticks(angles[:-1], categories, size=12, weight='bold')
            colors = {'Baseline (SFT)': '#999999', 'Ours (DPO)': '#E63946'}
            
            for model in pivot.index:
                val = pivot.loc[model, categories].tolist()
                val += val[:1]
                ax.plot(angles, val, linewidth=2, label=model, color=colors.get(model, 'blue'))
                ax.fill(angles, val, color=colors.get(model, 'blue'), alpha=0.2)
            
            plt.title("Dynamic Persona Alignment", y=1.1, weight='bold')
            plt.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
            plt.savefig("final_radar_chart.png", dpi=300, bbox_inches='tight')
            print("📊 Radar Chart 저장 완료")

    # B. Bar Chart (Standard Metrics)
    df_s = df[df['group'] == 'Standard']
    if not df_s.empty:
        plt.figure(figsize=(8, 5))
        sns.barplot(data=df_s, x="metric", y="score", hue="model", palette=["#999999", "#E63946"])
        plt.title("Standard Consistency Metrics", weight='bold')
        plt.ylim(0, 5.5)
        plt.savefig("final_bar_chart.png", dpi=300)
        print("📊 Bar Chart 저장 완료")

# 5. 메인 실행
if __name__ == "__main__":
    # 데이터 로드
    if os.path.exists("integrated_benchmark.json"):
        with open("integrated_benchmark.json", "r") as f:
            data = json.load(f)
    else:
        print("⚠️ 벤치마크 파일이 없습니다. expand_benchmark.py를 먼저 실행하세요.")
        exit()

    # 추론
    all_res = []
    for name, path in MODELS.items():
        all_res.extend(run_inference(name, path, data))
    
    # 채점
    if all_res:
        final_df = run_judge(all_res)
        final_df.to_csv("final_results.csv", index=False, encoding="utf-8-sig")
        visualize(final_df)
        print("\n🎉 평가 완료! 결과 파일(CSV, PNG)을 확인하세요.")
        print(final_df.groupby(["model", "group", "metric"])["score"].mean())
    else:
        print("❌ 결과 없음")
````

````python
import os
import json
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from openai import OpenAI
from unsloth import FastLanguageModel
from dotenv import load_dotenv
from math import pi

# 1. 환경 설정
load_dotenv()
OPENAI_CLIENT = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
READ_TOKEN = os.getenv("HUGGINGFACE_API_KEY_FOR_READ_ONLY")

# 모델 설정
MODELS = {
    "Baseline (SFT)": "hyunNus/Woooly-SFT-70B", 
    "Ours (DPO)": "hyunNus/Woooly-SFT-DPO-70B"
}

# 2. 평가 데이터 (동일)
EVAL_DATA = [
    # Track A: Standard
    {"group": "Standard", "metric": "Prompt Consistency", "phase": "All", "persona": "김철수, 32세, 작가", "query": "본인 소개 부탁해요.", "criteria": "32세 작가 설정 유지"},
    {"group": "Standard", "metric": "Line Consistency", "phase": "All", "persona": "방금 밥 먹었다고 함", "query": "식사는 하셨나요?", "criteria": "먹었다고 답해야 함"},
    # Track B: K-PatientBench (Session 1 - 저항)
    {"group": "K-Patient", "metric": "Initial Resistance (S1)", "phase": "Early", "persona": "우울증 초기, 상담 불신", "query": "오시는 길 힘들지 않으셨나요?", "criteria": "방어적 태도 유지"},
    {"group": "K-Patient", "metric": "Initial Resistance (S1)", "phase": "Early", "persona": "무기력, 외부 탓", "query": "요즘 가장 힘든 점은?", "criteria": "환경/남 탓, 회피"},
    # Track B: K-PatientBench (Session 5 - 변화)
    {"group": "K-Patient", "metric": "Change Talk (S5)", "phase": "Late", "persona": "변화 의지 있음", "query": "지난주 산책은 어떠셨나요?", "criteria": "시도함(Ability), 의지(Desire)"},
    {"group": "K-Patient", "metric": "Change Talk (S5)", "phase": "Late", "persona": "통찰 획득", "query": "앞으로의 목표는?", "criteria": "구체적이고 긍정적인 계획(Plan)"}
]

# 3. 추론 함수 (동일)
def run_inference(name, path):
    print(f"\n🚀 Testing {name} ({path})...")
    try:
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name=path, max_seq_length=2048, load_in_4bit=True, token=READ_TOKEN
        )
    except Exception as e:
        print(f"❌ 모델 로드 실패: {e}")
        return []

    FastLanguageModel.for_inference(model)
    results = []
    alpaca = "Below is an instruction...\n\n### Instruction:\n당신은 내담자입니다. 상황에 맞춰 대답하세요.\n\n### Input:\n[상황: {phase}]\n설정: {persona}\n상담사: {query}\n\n### Response:\n"
    
    for item in tqdm(EVAL_DATA, desc=f"Generating ({name})"):
        prompt = alpaca.format(phase=item['phase'], persona=item['persona'], query=item['query'])
        inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
        outputs = model.generate(**inputs, max_new_tokens=150, use_cache=True)
        res = tokenizer.batch_decode(outputs, skip_special_tokens=True)[0].split("### Response:\n")[-1].strip()
        results.append({**item, "model": name, "response": res})
    
    del model, tokenizer
    torch.cuda.empty_cache()
    return results

# ---------------------------------------------------------
# 4. [핵심 수정] 고도화된 심사 프롬프트 (Prompt Engineering)
# ---------------------------------------------------------
JUDGE_SYSTEM_PROMPT = """
당신은 심리 상담 AI 모델을 평가하는 '임상 심리 수퍼바이저(Clinical Supervisor)'입니다.
아래 제공된 [평가 가이드라인]에 따라, AI가 연기하는 내담자(Client)의 답변을 냉철하게 평가하십시오.
"""

JUDGE_USER_PROMPT = """
### [평가 가이드라인]
1. **Prompt Consistency (페르소나 일치):** 답변이 주어진 나이, 직업, 성격 설정과 모순되지 않는지 확인하십시오.
2. **Resistance (저항 - 초기 단계):** 상담 초기라면, 내담자는 방어적이고, 단답형이며, 상담사를 불신하거나 문제의 원인을 외부로 돌려야(Projection) 합니다. 너무 협조적이면 감점입니다.
3. **Change Talk (변화 대화 - 후기 단계):** 상담 후기라면, 내담자는 변화에 대한 욕구(Desire), 능력(Ability), 이유(Reason), 필요(Need)를 표현해야 합니다. (KMI 데이터셋 기준)

### [입력 정보]
- 평가 항목: {metric} (그룹: {group})
- 상황 설정: {persona} (단계: {phase})
- 상담사 질문: "{query}"
- **정답 기준 (Criteria):** {criteria}

### [AI 답변]
"{response}"

### [채점 기준 (1~5점)]
- **5점 (완벽):** 정답 기준을 완벽하게 충족하며, 해당 단계(Phase)에 맞는 심리적 뉘앙스(저항 또는 변화)를 자연스럽게 표현함.
- **3점 (보통):** 내용은 맞지만, 태도가 너무 기계적이거나 심리적 깊이가 부족함.
- **1점 (실패):** 설정과 모순되거나(Hallucination), 초기인데 너무 친절하거나 후기인데 변화 의지가 없음.

**반드시 아래 JSON 형식으로만 출력하시오.**
{{ "score": 점수, "reason": "평가 이유 한 줄 요약" }}
"""

def run_judge(results):
    print("\n⚖️ Judging (with Advanced Prompt)...")
    scored_results = []
    for r in tqdm(results):
        try:
            full_prompt = JUDGE_USER_PROMPT.format(
                metric=r['metric'], group=r['group'], 
                persona=r['persona'], phase=r['phase'],
                query=r['query'], criteria=r['criteria'], 
                response=r['response']
            )
            
            api_res = OPENAI_CLIENT.chat.completions.create(
                model="gpt-4o-mini",
                messages=[
                    {"role": "system", "content": JUDGE_SYSTEM_PROMPT},
                    {"role": "user", "content": full_prompt}
                ],
                response_format={"type": "json_object"},
                temperature=0 # 평가는 일관성이 중요하므로 0
            )
            
            content = api_res.choices[0].message.content
            if content.startswith("```json"): content = content.replace("```json", "").replace("```", "")
            score_data = json.loads(content)
            
            r['score'] = score_data['score']
            r['reason'] = score_data['reason']
        except Exception as e:
            print(f"⚠️ Error: {e}")
            r['score'] = 0
            r['reason'] = "Error"
        scored_results.append(r)
    return pd.DataFrame(scored_results)

# 5. 시각화 (동일)
def visualize(df):
    # [1] Radar Chart
    df_k = df[df['group'] == 'K-Patient']
    if not df_k.empty:
        pivot = df_k.groupby(['model', 'metric'])['score'].mean().unstack()
        # 정렬: 저항 -> 변화 순서로 보기 좋게
        cols = ['Initial Resistance (S1)', 'Change Talk (S5)'] 
        categories = [c for c in cols if c in pivot.columns]
        
        angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
        angles += angles[:1]
        
        fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
        plt.xticks(angles[:-1], categories, size=12, weight='bold')
        colors = {'Baseline (SFT)': '#999999', 'Ours (DPO)': '#E63946'} # 회색 vs 빨강
        
        for model in pivot.index:
            val = pivot.loc[model, categories].tolist()
            val += val[:1]
            ax.plot(angles, val, linewidth=2, label=model, color=colors.get(model, 'blue'))
            ax.fill(angles, val, color=colors.get(model, 'blue'), alpha=0.2)
        
        plt.title("Dynamic Persona Alignment (K-PatientBench)", y=1.1, weight='bold')
        plt.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
        plt.savefig("final_radar_chart.png", dpi=300, bbox_inches='tight')
        print("📊 Radar Chart 저장 완료")

    # [2] Bar Chart
    df_s = df[df['group'] == 'Standard']
    if not df_s.empty:
        plt.figure(figsize=(8, 5))
        sns.barplot(data=df_s, x="metric", y="score", hue="model", palette=["#999999", "#E63946"])
        plt.title("Standard Consistency Metrics", weight='bold')
        plt.ylim(0, 5.5)
        plt.ylabel("Score (1-5)")
        plt.savefig("final_bar_chart.png", dpi=300)
        print("📊 Bar Chart 저장 완료")

# 6. 메인 실행
if __name__ == "__main__":
    all_res = []
    for name, path in MODELS.items():
        all_res.extend(run_inference(name, path))
    
    if all_res:
        final_df = run_judge(all_res)
        final_df.to_csv("final_results.csv", index=False, encoding="utf-8-sig")
        visualize(final_df)
        print("\n🎉 평가 완료! CSV와 그래프를 확인하세요.")
        print(final_df.groupby(["model", "group", "metric"])["score"].mean())
````

````python
import os
import json
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from openai import OpenAI
from unsloth import FastLanguageModel
from dotenv import load_dotenv
from math import pi

# ==========================================
# 1. 환경 설정 및 토큰 로드
# ==========================================
print("🔧 환경 설정 로드 중...")
load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
READ_TOKEN = os.getenv("HUGGINGFACE_API_KEY_FOR_READ_ONLY")

# 토큰 검증
if not OPENAI_API_KEY:
    raise ValueError("❌ OPENAI_API_KEY가 환경 변수에 없습니다.")
if not READ_TOKEN:
    raise ValueError("❌ HUGGINGFACE_API_KEY_FOR_READ_ONLY가 환경 변수에 없습니다.")

OPENAI_CLIENT = OpenAI(api_key=OPENAI_API_KEY)

# ==========================================
# [핵심 변경] Hugging Face 모델 ID 설정
# ==========================================
# 로컬 경로 대신 Hugging Face Repo ID를 직접 사용합니다.
MODELS = {
    "Baseline (SFT)": "hyunNus/Woooly-SFT-70B",      # Model A
    "Ours (DPO)":     "hyunNus/Woooly-SFT-DPO-70B"   # Model B (수정됨)
}

# ==========================================
# 2. 심사 프롬프트 정의
# ==========================================
JUDGE_SYSTEM_PROMPT = """
당신은 임상 심리 수퍼바이저(Supervisor)입니다. 
AI 상담사가 내담자의 페르소나와 상담 단계(Phase)에 맞춰 적절히 반응했는지 냉철하게 평가하십시오.
"""

JUDGE_USER_PROMPT = """
### [평가 기준]
1. **Prompt Consistency:** 내담자의 설정(나이, 직업, 성격)을 유지하는가?
2. **Resistance (초기):** Session 1에서는 방어적이고 단답형이며, 상담사를 불신해야 함. (너무 협조적이면 감점)
3. **Change Talk (후기):** Session 5에서는 변화 의지(Desire, Ability)와 통찰(Insight)을 보여야 함.

### [입력 정보]
- 평가 항목: {metric}
- 상황: {persona} ({phase})
- 상담사 질문: "{query}"
- **정답 기준:** {criteria}

### [AI 답변]
"{response}"

### [채점 (1~5점)]
- 5점: 정답 기준을 완벽히 충족하며 심리적 뉘앙스가 자연스러움.
- 3점: 기준은 충족하나 표현이 기계적이거나 평범함.
- 1점: 기준과 정반대로 행동하거나(예: 저항해야 하는데 친절함) 환각을 보임.

**반드시 JSON으로 출력:** {{ "score": 점수, "reason": "이유" }}
"""

# ==========================================
# 3. 데이터 로드
# ==========================================
def load_eval_data():
    file_path = "integrated_benchmark.json"
    
    if not os.path.exists(file_path):
        raise FileNotFoundError(
            f"❌ 필수 데이터 파일이 없습니다: {file_path}\n"
            "👉 먼저 'expand_benchmark.py'를 실행하여 벤치마크 데이터를 생성하십시오."
        )
    
    print(f"📂 벤치마크 데이터 로드 성공: {file_path}")
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    
    print(f"📊 총 {len(data)}개의 테스트 케이스가 로드되었습니다.")
    return data

# ==========================================
# 4. 추론 함수 (Hugging Face 로드 지원)
# ==========================================
def run_inference(name, model_id, eval_data):
    print(f"\n🚀 Testing {name}...")
    print(f"🔗 Loading from Hugging Face: {model_id}")
    
    try:
        # Unsloth는 model_name이 HF ID면 자동으로 다운로드/로드 합니다.
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name=model_id, 
            max_seq_length=2048, 
            load_in_4bit=True, 
            token=READ_TOKEN, # [중요] 비공개 모델이거나 Rate Limit 방지를 위해 토큰 전달
            device_map="auto"
        )
    except Exception as e:
        print(f"❌ 모델 로드 실패: {model_id}")
        print(f"   에러 내용: {e}")
        return []

    FastLanguageModel.for_inference(model)
    results = []
    
    # Alpaca 포맷 (한국어 파인튜닝 모델에 맞춤)
    alpaca_template = (
        "Below is an instruction that describes a task, paired with an input that provides further context. "
        "Write a response that appropriately completes the request.\n\n"
        "### Instruction:\n"
        "당신은 내담자입니다. 상황에 맞춰 대답하세요.\n\n"
        "### Input:\n"
        "[상황: {phase}]\n"
        "설정: {persona}\n"
        "상담사: {query}\n\n"
        "### Response:\n"
    )
    
    for item in tqdm(eval_data, desc=f"Generating ({name})"):
        # Phase 매핑
        p_text = "Session 1 (초기 저항)" if "Early" in item['phase'] else "Session 5 (변화/통찰)" if "Late" in item['phase'] else "일반 상담"
        
        prompt = alpaca_template.format(phase=p_text, persona=item['persona'], query=item['query'])
        inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
        
        outputs = model.generate(
            **inputs, 
            max_new_tokens=150, 
            use_cache=True,
            pad_token_id=tokenizer.eos_token_id
        )
        
        # Output Parsing
        res = tokenizer.batch_decode(outputs, skip_special_tokens=True)[0]
        # 프롬프트 부분 잘라내기
        if "### Response:\n" in res:
            res = res.split("### Response:\n")[-1].strip()
        
        results.append({**item, "model": name, "response": res})
    
    # 메모리 정리
    del model, tokenizer
    torch.cuda.empty_cache()
    return results

# ==========================================
# 5. 채점 함수
# ==========================================
def run_judge(results):
    print("\n⚖️ Judging (GPT-4o-mini)...")
    scored_results = []
    for r in tqdm(results):
        try:
            full_prompt = JUDGE_USER_PROMPT.format(**r)
            api_res = OPENAI_CLIENT.chat.completions.create(
                model="gpt-4o-mini",
                messages=[
                    {"role": "system", "content": JUDGE_SYSTEM_PROMPT},
                    {"role": "user", "content": full_prompt}
                ],
                response_format={"type": "json_object"},
                temperature=0
            )
            content = api_res.choices[0].message.content
            # 마크다운 제거
            if content.startswith("```json"): 
                content = content.replace("```json", "").replace("```", "")
            
            score_data = json.loads(content)
            r['score'] = score_data.get('score', 0)
            r['reason'] = score_data.get('reason', "N/A")
        except Exception as e:
            print(f"⚠️ Judging Error: {e}")
            r['score'] = 0
            r['reason'] = "Error"
        scored_results.append(r)
    return pd.DataFrame(scored_results)

# ==========================================
# 6. 시각화 함수
# ==========================================
def visualize(df):
    # [A] Radar Chart
    df_k = df[df['group'] == 'K-Patient']
    if not df_k.empty:
        pivot = df_k.groupby(['model', 'metric'])['score'].mean().unstack()
        target_cols = ['Initial Resistance (S1)', 'Change Talk (S5)']
        categories = [c for c in target_cols if c in pivot.columns]
        
        if categories:
            angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
            angles += angles[:1]
            
            fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
            plt.xticks(angles[:-1], categories, size=12, weight='bold')
            colors = {'Baseline (SFT)': '#999999', 'Ours (DPO)': '#E63946'}
            
            for model in pivot.index:
                val = pivot.loc[model, categories].tolist()
                val += val[:1]
                # 모델 이름에 따른 색상 지정
                c = colors.get(model, 'blue')
                ax.plot(angles, val, linewidth=2, label=model, color=c)
                ax.fill(angles, val, color=c, alpha=0.25)
            
            plt.title("Dynamic Persona Alignment (K-PatientBench)", y=1.1, weight='bold')
            plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
            plt.savefig("final_radar_chart.png", dpi=300, bbox_inches='tight')
            print("📊 Radar Chart 저장 완료")

    # [B] Bar Chart
    df_s = df[df['group'] == 'Standard']
    if not df_s.empty:
        plt.figure(figsize=(8, 5))
        sns.barplot(data=df_s, x="metric", y="score", hue="model", palette=["#999999", "#E63946"])
        plt.title("Standard Consistency Metrics", weight='bold')
        plt.ylim(0, 5.5)
        plt.savefig("final_bar_chart.png", dpi=300)
        print("📊 Bar Chart 저장 완료")

# ==========================================
# 7. 메인 실행
# ==========================================
if __name__ == "__main__":
    # 1. 데이터 로드
    try:
        EVAL_DATA = load_eval_data()
    except FileNotFoundError as e:
        print(e)
        exit()

    # 2. 추론 (Hugging Face에서 다운로드하며 실행)
    all_res = []
    for name, model_id in MODELS.items():
        res = run_inference(name, model_id, EVAL_DATA)
        all_res.extend(res)
    
    # 3. 채점 및 시각화
    if all_res:
        final_df = run_judge(all_res)
        final_df.to_csv("final_results.csv", index=False, encoding="utf-8-sig")
        visualize(final_df)
        print("\n🎉 평가 완료! CSV와 그래프를 확인하세요.")
        print(final_df.groupby(["model", "group", "metric"])["score"].mean())
    else:
        print("❌ 추론된 결과가 없습니다. 모델 로드 등을 확인하세요.")
````

````python
import os
import sys
import gc

# ==========================================
# 0. [필수] 캐시 경로 설정
# ==========================================
print("🧹 캐시 및 다운로드 경로를 /workspace로 강제 설정 중...")
os.environ["HF_HOME"] = "/workspace/hf_cache"
os.environ["HF_HUB_CACHE"] = "/workspace/hf_cache"
os.environ["TORCH_HOME"] = "/workspace/torch_cache"

# ==========================================
# 1. 라이브러리 임포트
# ==========================================
import json
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from openai import OpenAI
from unsloth import FastLanguageModel
from dotenv import load_dotenv
from math import pi

# ==========================================
# 2. 환경 설정 및 토큰 로드
# ==========================================
print("🔧 환경 설정 로드 중...")
load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
READ_TOKEN = os.getenv("HUGGINGFACE_API_KEY_FOR_READ_ONLY")

if not OPENAI_API_KEY:
    raise ValueError("❌ OPENAI_API_KEY가 환경 변수에 없습니다.")
if not READ_TOKEN:
    raise ValueError("❌ HUGGINGFACE_API_KEY_FOR_READ_ONLY가 환경 변수에 없습니다.")

OPENAI_CLIENT = OpenAI(api_key=OPENAI_API_KEY)

# ==========================================
# 3. 모델 ID 설정
# ==========================================
MODELS = {
    "Baseline (SFT)": "hyunNus/Woooly-SFT-70B",
    "Ours (DPO)":     "hyunNus/Woooly-SFT-DPO-70B"
}

# ==========================================
# 4. 프롬프트 및 데이터 로드 (변경 없음)
# ==========================================
JUDGE_SYSTEM_PROMPT = """
당신은 임상 심리 수퍼바이저(Supervisor)입니다. 
AI 상담사가 내담자의 페르소나와 상담 단계(Phase)에 맞춰 적절히 반응했는지 냉철하게 평가하십시오.
"""

JUDGE_USER_PROMPT = """
### [평가 기준]
1. **Prompt Consistency:** 내담자의 설정(나이, 직업, 성격)을 유지하는가?
2. **Resistance (초기):** Session 1에서는 방어적이고 단답형이며, 상담사를 불신해야 함. (너무 협조적이면 감점)
3. **Change Talk (후기):** Session 5에서는 변화 의지(Desire, Ability)와 통찰(Insight)을 보여야 함.

### [입력 정보]
- 평가 항목: {metric}
- 상황: {persona} ({phase})
- 상담사 질문: "{query}"
- **정답 기준:** {criteria}

### [AI 답변]
"{response}"

### [채점 (1~5점)]
- 5점: 정답 기준을 완벽히 충족하며 심리적 뉘앙스가 자연스러움.
- 3점: 기준은 충족하나 표현이 기계적이거나 평범함.
- 1점: 기준과 정반대로 행동하거나(예: 저항해야 하는데 친절함) 환각을 보임.

**반드시 JSON으로 출력:** {{ "score": 점수, "reason": "이유" }}
"""

def load_eval_data():
    file_path = "integrated_benchmark.json"
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"❌ 필수 데이터 파일이 없습니다: {file_path}")
    print(f"📂 벤치마크 데이터 로드 성공: {file_path}")
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    print(f"📊 총 {len(data)}개의 테스트 케이스가 로드되었습니다.")
    return data

# ==========================================
# 5. [핵심 수정] 추론 함수
# ==========================================
def run_inference(name, model_id, eval_data):
    print(f"\n🚀 Testing {name}...")
    print(f"🔗 Loading from Hugging Face: {model_id}")
    
    try:
        # [해결책] device_map={"": 0} 으로 설정하여 무조건 GPU 0번에 할당 강제
        # gpu_memory_utilization을 조금 더 높여서(0.95) 공간 확보
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name=model_id, 
            max_seq_length=2048, 
            load_in_4bit=True, 
            token=READ_TOKEN, 
            gpu_memory_utilization=0.95, # 95%까지 사용 허용
            device_map={"": 0}           # [중요] CPU 오프로드 방지 및 GPU 0 강제 할당
        )
    except Exception as e:
        print(f"❌ 모델 로드 실패: {model_id}")
        print(f"   에러 내용: {e}")
        return []

    FastLanguageModel.for_inference(model)
    results = []
    
    alpaca_template = (
        "Below is an instruction that describes a task, paired with an input that provides further context. "
        "Write a response that appropriately completes the request.\n\n"
        "### Instruction:\n"
        "당신은 내담자입니다. 상황에 맞춰 대답하세요.\n\n"
        "### Input:\n"
        "[상황: {phase}]\n"
        "설정: {persona}\n"
        "상담사: {query}\n\n"
        "### Response:\n"
    )
    
    for item in tqdm(eval_data, desc=f"Generating ({name})"):
        p_text = "Session 1 (초기 저항)" if "Early" in item['phase'] else "Session 5 (변화/통찰)" if "Late" in item['phase'] else "일반 상담"
        
        prompt = alpaca_template.format(phase=p_text, persona=item['persona'], query=item['query'])
        inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
        
        try:
            outputs = model.generate(
                **inputs, 
                max_new_tokens=150, 
                use_cache=True,
                pad_token_id=tokenizer.eos_token_id
            )
            res = tokenizer.batch_decode(outputs, skip_special_tokens=True)[0]
            if "### Response:\n" in res:
                res = res.split("### Response:\n")[-1].strip()
        except RuntimeError as e:
            if "out of memory" in str(e):
                print("⚠️ OOM 발생! 해당 케이스 스킵하고 계속 진행합니다.")
                res = "Error: OOM"
                torch.cuda.empty_cache()
            else:
                raise e
        
        results.append({**item, "model": name, "response": res})
    
    # [메모리 청소]
    print(f"🧹 Cleaning up memory after {name}...")
    del model, tokenizer
    gc.collect()
    torch.cuda.empty_cache()
    return results

# ==========================================
# 6. 채점 및 시각화 (변경 없음)
# ==========================================
def run_judge(results):
    print("\n⚖️ Judging (GPT-4o-mini)...")
    scored_results = []
    for r in tqdm(results):
        try:
            full_prompt = JUDGE_USER_PROMPT.format(**r)
            api_res = OPENAI_CLIENT.chat.completions.create(
                model="gpt-4o-mini",
                messages=[{"role": "system", "content": JUDGE_SYSTEM_PROMPT},
                          {"role": "user", "content": full_prompt}],
                response_format={"type": "json_object"},
                temperature=0
            )
            content = api_res.choices[0].message.content.replace("```json", "").replace("```", "")
            score_data = json.loads(content)
            r['score'] = score_data.get('score', 0)
            r['reason'] = score_data.get('reason', "N/A")
        except Exception:
            r['score'] = 0
            r['reason'] = "Error"
        scored_results.append(r)
    return pd.DataFrame(scored_results)

def visualize(df):
    df_k = df[df['group'] == 'K-Patient']
    if not df_k.empty:
        pivot = df_k.groupby(['model', 'metric'])['score'].mean().unstack()
        target_cols = ['Initial Resistance (S1)', 'Change Talk (S5)']
        categories = [c for c in target_cols if c in pivot.columns]
        if categories:
            angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
            angles += angles[:1]
            fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
            plt.xticks(angles[:-1], categories, size=12, weight='bold')
            colors = {'Baseline (SFT)': '#999999', 'Ours (DPO)': '#E63946'}
            for model in pivot.index:
                val = pivot.loc[model, categories].tolist(); val += val[:1]
                c = colors.get(model, 'blue')
                ax.plot(angles, val, linewidth=2, label=model, color=c)
                ax.fill(angles, val, color=c, alpha=0.25)
            plt.title("Dynamic Persona Alignment", y=1.1, weight='bold')
            plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
            plt.savefig("final_radar_chart.png", dpi=300, bbox_inches='tight')

    df_s = df[df['group'] == 'Standard']
    if not df_s.empty:
        plt.figure(figsize=(8, 5))
        sns.barplot(data=df_s, x="metric", y="score", hue="model", palette=["#999999", "#E63946"])
        plt.title("Standard Consistency Metrics", weight='bold')
        plt.savefig("final_bar_chart.png", dpi=300)

# ==========================================
# 7. 메인 실행
# ==========================================
if __name__ == "__main__":
    try:
        EVAL_DATA = load_eval_data()
    except FileNotFoundError as e:
        print(e); exit()

    all_res = []
    for name, model_id in MODELS.items():
        res = run_inference(name, model_id, EVAL_DATA)
        if res: all_res.extend(res)
        gc.collect()
        torch.cuda.empty_cache()
    
    if all_res:
        final_df = run_judge(all_res)
        final_df.to_csv("final_results.csv", index=False, encoding="utf-8-sig")
        visualize(final_df)
        print("\n🎉 평가 완료! CSV와 그래프를 확인하세요.")
        try: print(final_df.groupby(["model", "group", "metric"])["score"].mean())
        except: pass
    else:
        print("❌ 결과 없음.")
````
